```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('iris.csv')
```

```python
df.info()
```

```python
plt.figure(figsize=(10, 6))
species_list = df['species'].unique()
colors = ['red', 'blue', 'green']

for i, species in enumerate(species_list):
    species_data = df[df['species'] == species]
    plt.scatter(species_data['sepal_length'], species_data['sepal_width'], 
               c=colors[i], label=species, alpha=0.7)

plt.xlabel('Sepal Length (cm)')
plt.ylabel('Sepal Width (cm)')
plt.title('Sepal Length vs Sepal Width by Species')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
species_sepal_length_mean = df.groupby('species')['sepal_length'].mean()

bars = plt.bar(species_sepal_length_mean.index, species_sepal_length_mean.values, 
               color=['skyblue', 'lightcoral', 'lightgreen'], alpha=0.8)

# 막대 위에 수치 표시
for bar, value in zip(bars, species_sepal_length_mean.values):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.05, 
             f'{value:.2f}', ha='center', va='bottom', fontweight='bold')

plt.xlabel('Species')
plt.ylabel('Average Sepal Length (cm)')
plt.title('Average Sepal Length by Species')
plt.xticks(rotation=45)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
from sklearn import datasets
iris = datasets.load_iris()
print('iris 키:', iris.keys())
```

```python
print('데이터:\n', iris['data'][:5])          # 샘플 5개
print('특성 이름:', iris['feature_names'])
print('타깃:\n', iris['target'][:5])
print('타깃 이름:', iris['target_names'])
print('데이터 크기:', iris['data'].shape)
print('타깃 크기:', iris['target'].shape)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(iris['data'], iris['target'])
```

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=3)    # 모델 설정
knn.fit(X_train, y_train)                    # 학습
```

```python
X_new = [[6, 3.5, 5, 1.5]]            # 꽃받침 길이 6, 너비 3.5, 꽃잎 길이 5, 너비 1.5
pred = knn.predict(X_new)
print('예측 값:', pred, '->', iris['target_names'][pred])
```

```python
y_pred = knn.predict(X_test)
print('테스트 세트 예측 값:\n', y_pred)
print('테스트 세트 실제 타깃:\n', y_test)
```

```python
print(f'훈련 세트 정확도: {knn.score(X_train, y_train):.3f}')
print(f'테스트 세트 정확도: {knn.score(X_test, y_test):.3f}')
```

```python
# -*- coding: utf-8 -*-
# 실습 8. 회귀분석의 개념: 기온과 아이스크림 판매량 [그림 13-19 ~ 13-21]
# 교재의 원자료는 슬라이드 그림이라 비슷한 경향의 연습용 데이터를 직접 만들었다.
# 그래서 아래 회귀식의 숫자는 교재의 '35.9 + 2.9 x 평균 기온' 과 조금 다르다.
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

rng = np.random.default_rng(0)
temp = np.round(rng.uniform(15, 35, 20), 1)               # 일간 평균 기온
sales = np.round(36 + 2.9 * temp + rng.normal(0, 6, 20))  # 판매량 = 직선 + 잡음

X = temp.reshape(-1, 1)          # 특성은 2차원 배열(20행 1열)이어야 한다
lr = LinearRegression()
lr.fit(X, sales)

print(f'절편: {lr.intercept_:.2f}')
print(f'계수: {lr.coef_[0]:.2f}')
print(f'회귀식: 판매량 = {lr.intercept_:.1f} + {lr.coef_[0]:.1f} x 기온')
print('기온 26도의 예측 판매량:', lr.predict([[26]]).round(1))

plt.figure(figsize=(6, 4))
plt.scatter(temp, sales, label='data')                         # (a) 산점도
plt.plot(np.sort(temp), lr.predict(np.sort(temp).reshape(-1, 1)),
         color='red', label='regression line')                 # (b) 대표 직선
plt.xlabel('temperature')
plt.ylabel('ice cream sales')
plt.legend()
plt.savefig('lab08_icecream.png', dpi=120, bbox_inches='tight')
plt.close()
```
