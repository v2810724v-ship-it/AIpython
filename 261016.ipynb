```python
# 1부터 100까지의 합을 출력

sum = 0
for a in range(101):
  sum += a
print(sum)
```

```python
# 과제 4-1-1

sum = 0
for a in range(2,101,2):
  sum += a
print(sum)
```

```python
# 과제 4-1-1 (if문 사용)

sum = 0
for a in range(1, 101):
    if a % 2 == 0:
        sum += a
print(sum)
```

```python
sum = 0
a = 1
while a<=100:
  sum += a
  a += 1
print(sum)
```

```python
sum = 0
i = 1
while True:
  sum += i
  i += 1
  if i > 100:
    break
print(sum)
```

```python
# 양의 정수를 계속 입력 받아 합계를 구하고 0을 입력하면 반복을 끝내고 합계를 출력

sum = 0
while True:
  n = int(input('양의 정수를 입력하시오: '))
  if n == 0:
    break
  sum += n
print(sum)
```

```python
max_score = 0
scores = [85, 95, 70, 75, 99, 93, 88]
for i in scores:
  if i > max_score:
    max_score = i
print(max_score)
```
