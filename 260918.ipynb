```python
# 음료 가격
americano_price = 3000
latte_price = 4500
green_tea_price = 5000

total = 0

americano_num = int(input("아메리카노 주문 수: "))
latte_num = int(input("라떼 주문 수: "))
green_tea_num = int(input("녹차 주문 수: "))

total += americano_price * americano_num
total += latte_price * latte_num
total += green_tea_price * green_tea_num

print(f"전체 주문 금액은 {total} 원입니다.")
```
